```python
def mediana(data): # 10 => 5,6 
    no_elem = len(data)
    index_mid = no_elem // 2
    if no_elem % 2 == 0: # numarul de elemente este par
        # no_elem == 10 => index_mid = 10/2 = 5
        return (data[index_mid-1] + data[index_mid]) / 2
    else: # numarul de elemente este impar
        # no_elem == 9
        return data[index_mid]
```

```python
def percentile(data, p):
    if p == 50:
        return mediana(data)
    if p == 25:
        return mediana(data[:(len(data)//2)])
    if p == 75:
        return mediana(data[(len(grades)//2):])
    if p == 100:
        return data[len(data)-1]
    return None
```

```python
mediana(grades)
```

```python
mediana(grades_impar)
```
