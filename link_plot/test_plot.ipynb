```python
streaming_platforms = ['Netflix', 'Disney+', 'YouTube', 'Amazon Prime', 'Spotify']
users = [230, 150, 270, 200, 240]

plt.figure(figsize=(12.80, 10.24)) # 1280 x 1024

plt.bar(streaming_platforms, users, 0.7)

plt.title('Numar utiliatori platforme streaming (mil.)')
plt.xlabel('Platforma', loc='right')
plt.ylabel('Utilizatori (mil.)')
```

```python
zile = ['Lu', 'Ma', 'Mi', 'Jo', 'Vi', 'Sa', 'Du']
temp = [18, 20, 21, 19, 23, 26, 24]

plt.figure(figsize=(10, 5)) # w = 10 * 100 (dpi) , h = 5 * 100 (dpi) => 1000px x 500px
plt.plot(zile, temp)
plt.ylim(bottom=0)

plt.title('Evolutia temperaturii in timpul saptamanii')
plt.xlabel('Zi')
plt.ylabel('Temperatura (C)')
```
