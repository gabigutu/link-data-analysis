# Întrebarea I - Care este produsul cu cele mai mari vânzări?

```python
df_aux = df[ ['product', 'price', 'quantity', 'revenue'] ]
df_aux
```

```python
df_sales = df_aux.groupby('product').sum()
df_sales
```

```python
sales_x = list(df_sales.index)
sales_heights = df_sales['quantity'].values
```

```python
plt.bar(sales_x, sales_heights)
```
