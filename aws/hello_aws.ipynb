```python
BUCKET_TO_CREATE = 'galeata.mea'
```

```python
import boto3
import os
import pandas as pd
import s3fs
from backports import configparser
```

```python
config = configparser.RawConfigParser()
config.read('../keys/aws.ini')
```

```python
os.environ['AWS_ACCESS_KEY_ID'] = config.get('DEFAULT', 'AWS_ACCESS_KEY_ID')
os.environ['AWS_SECRET_ACCESS_KEY'] = config.get('DEFAULT', 'AWS_SECRET_ACCESS_KEY')
```

```python
s3_client = boto3.client('s3')
type(s3_client)
```

```python
s3_client.create_bucket(Bucket=BUCKET_TO_CREATE)
print(f'Bucket-ul urmator a fost creat: {BUCKET_TO_CREATE}')
```

```python
s3_client.upload_file('../data/customers-1000.csv', BUCKET_TO_CREATE, 'raw/customers-100.csv')
```

```python
cust_100_dict = s3_client.get_object(Bucket=BUCKET_TO_CREATE, Key='raw/customers-100.csv')
cust_100_dict['Body'].read().decode('utf-8')
```
